This default notebook is executed using Databricks Workflows as defined in resources/Databricks_TableCreation.job.yml.

```python
from pyspark.sql import SparkSession

# Initialize Spark
spark = SparkSession.builder.appName("NYC Taxi Data").getOrCreate()

# Create DataFrame
data = [
    ("trip_001", "2025-03-01", 12.50, "Credit Card"),
    ("trip_002", "2025-03-02", 8.75, "Cash"),
    ("trip_003", "2025-03-02", 20.00, "Credit Card"),
    ("trip_004", "2025-03-03", 15.30, "Debit Card"),
]

columns = ["trip_id", "date", "fare_amount", "payment_type"]

df = spark.createDataFrame(data, columns)

# Write to Delta Table
df.write.format("delta").mode("overwrite").saveAsTable("nyc_taxi_data")
```
